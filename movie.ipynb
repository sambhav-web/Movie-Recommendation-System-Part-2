```python
df=df[df['genres']!='[]']
```

```python
import ast
def handle_genres(col):
  L=[]
  for i in ast.literal_eval(col):
    L.append(i['name'])
  return L
df['genres']=df['genres'].apply(handle_genres)
```

```python
df=df[df['cast']!='[]']
```

```python
df['cast'][0]
```

```python
def handle_cast(col):
  L=[]
  j=0
  for i in ast.literal_eval(col):
    if j<5:
      L.append(i['name'])
    j+=1
  return L
df['cast']=df['cast'].apply(handle_cast)
```

## We can't just go and delete these many records. Also we need keyword column as important feature. So let's make it for our use.

```python
df['keywords'][0]
```

```python
def handle_keywords(col):
  L=[]
  for i in ast.literal_eval(col):
      if i is not None:
        L.append(i['name'])
  return L
df['keywords']=df['keywords'].apply(handle_keywords)
```

```python
df=df[df['crew']!='[]']
```

```python
df['crew'][3]
```

```python
def handle_crew(col):
  L=[]
  for i in ast.literal_eval(col):
    if i['job']=='Director':
      L.append(i['name'])
      break
  return L
df['crew']=df['crew'].apply(handle_crew)
```

```python
df.reset_index(inplace=True,drop=True)
```

```python
df['overview']=df['overview'].apply(lambda x:x.lower())
```

```python
import re
def cleaner(col):
  col=re.sub('<.*?>','',col)
  col=re.sub('https?://\S+|www\.\S+','',col)
  return re.sub('[^\w\s]','',col)
df['overview']=df['overview'].apply(cleaner)
```

```python
import nltk
nltk.download('stopwords')
```

```python
from nltk.tokenize import wordpunct_tokenize
from nltk.stem import SnowballStemmer
from nltk.corpus import stopwords
stemmer=SnowballStemmer('english')
words=stopwords.words('english')
```

```python
def stopword_removal(col):
  L=[]
  tokens=wordpunct_tokenize(col)
  for i in tokens:
    if i not in words:
      L.append(i)
  return " ".join(L)
df['overview']=df['overview'].apply(stopword_removal)
```

```python
def stemming(col):
  L=[]
  tokens=wordpunct_tokenize(col)
  for i in tokens:
    L.append(stemmer.stem(i))
  return L
df['overview']=df['overview'].apply(stemming)
```

## We need to lowercase all the columns before merging them and along make the entire name as the single entity

```python
df['Collection']=df['cast']+df['crew']+df['genres']+df['keywords']
```

```python
df.sample(1)
```

```python
df['Collection'][0]
```

```python
def final_touch(col):
  L=[]
  for i in col:
    L.append("".join(i.lower().split()))
  return L
df['Collection']=df['Collection'].apply(final_touch)
```

```python
df['tags']=df['Collection']+df['overview']
```

```python
df['tags']=df['tags'].apply(lambda x: " ".join(x))
```
